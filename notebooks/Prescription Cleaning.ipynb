```python
import pandas as pd

# --- 1. Load your comprehensive LSOA lookup file ---
lsoa_lookup_path = '/Users/callumhall/Documents/Dissertation/Datasets/PCD to LSOA to MSOA to LAD.csv'

# THE FIX IS HERE: Add the encoding='latin1' parameter
df_lookup = pd.read_csv(
    lsoa_lookup_path,
    usecols=['ladnm', 'pcds'],
    encoding='latin1'  # <--- ADD THIS
)

# --- 2. Create a list of all London Borough names ---
london_boroughs = [
    'Camden', 'Greenwich', 'Hackney', 'Hammersmith and Fulham', 'Islington', 'Kensington and Chelsea',
    'Lambeth', 'Lewisham', 'Southwark', 'Tower Hamlets', 'Wandsworth', 'Westminster',
    'Barking and Dagenham', 'Barnet', 'Bexley', 'Brent', 'Bromley', 'Croydon', 'Ealing', 'Enfield',
    'Haringey', 'Harrow', 'Havering', 'Hillingdon', 'Hounslow', 'Kingston upon Thames', 'Merton',
    'Newham', 'Redbridge', 'Richmond upon Thames', 'Sutton', 'Waltham Forest',
    'City of London'
]

# --- 3. Filter the lookup to keep only London postcodes ---
df_london_postcodes = df_lookup[df_lookup['ladnm'].isin(london_boroughs)]

# --- 4. Create a Python 'set' of London postcodes for fast filtering ---
london_postcodes_set = set(df_london_postcodes['pcds'])

print(f"Created a reference list of {len(london_postcodes_set)} postcodes in London.")
```

```python
def process_monthly_prescription_file(input_path, output_path, london_postcodes):
    """
    Loads a single monthly prescription CSV, filters it for London and mental health drugs,
    aggregates it by GP practice, and saves the result to a new CSV.
    """
    try:
        print(f"Processing {input_path}...")
        
        # 1. Load the raw monthly data
        df_monthly = pd.read_csv(input_path)
        
        # 2. Filter for London practices using the postcode set
        df_london = df_monthly[df_monthly['POSTCODE'].isin(london_postcodes)].copy()
        
        # 3. Filter for mental health prescriptions (Anxiolytics & Antidepressants)
        df_london['BNF_CODE'] = df_london['BNF_CODE'].astype(str)
        is_anxiolytic = df_london['BNF_CODE'].str.startswith('0401')
        is_antidepressant = df_london['BNF_CODE'].str.startswith('0403')
        df_mh = df_london[is_anxiolytic | is_antidepressant]
        
        # 4. Aggregate the data by GP Practice
        df_summary = df_mh.groupby(['PRACTICE_CODE', 'POSTCODE'])['ITEMS'].sum().reset_index()
        df_summary.rename(columns={'ITEMS': 'Total_MH_Items'}, inplace=True)
        
        # 5. Save the cleaned, aggregated data to a new file
        df_summary.to_csv(output_path, index=False)
        
        print(f"Successfully processed and saved to {output_path}")
        
    except Exception as e:
        print(f"An error occurred while processing {input_path}: {e}")
```

```python
import pandas as pd

# --- 1. Load your LSOA to Higher Authority Lookup ---
# This file links LSOAs to Boroughs (LADs)
lookup_path = '/Users/callumhall/Documents/Dissertation/Datasets/PCD to LSOA to MSOA to LAD.csv' #<-- Update this path
df_lookup = pd.read_csv(lookup_path, 
                        usecols=['ladcd', 'ladnm'],
                       encoding = 'latin1'
                       )

# --- 2. Get a definitive list of London Borough CODES ---
# We filter for the 33 London Boroughs and get their unique codes
london_boroughs = [
    'Camden', 'Greenwich', 'Hackney', 'Hammersmith and Fulham', 'Islington', 'Kensington and Chelsea',
    'Lambeth', 'Lewisham', 'Southwark', 'Tower Hamlets', 'Wandsworth', 'Westminster',
    'Barking and Dagenham', 'Barnet', 'Bexley', 'Brent', 'Bromley', 'Croydon', 'Ealing', 'Enfield',
    'Haringey', 'Harrow', 'Havering', 'Hillingdon', 'Hounslow', 'Kingston upon Thames', 'Merton',
    'Newham', 'Redbridge', 'Richmond upon Thames', 'Sutton', 'Waltham Forest',
    'City of London'
]
london_lad_codes = df_lookup[df_lookup['ladnm'].isin(london_boroughs)]['ladcd'].unique()

# --- 3. Load and Filter HPI Data by Borough Code ---
hpi_path = '/Users/callumhall/Documents/Dissertation/Datasets/HPI Data.csv'
cols_to_use = ['Date','RegionName', 'AreaCode', 'AveragePrice']
df_hpi_raw = pd.read_csv(hpi_path, usecols=cols_to_use)
df_hpi_raw['Date'] = pd.to_datetime(df_hpi_raw['Date'], format='%m/%d/%Y')
df_hpi_filtered = df_hpi_raw[df_hpi_raw['Date'].dt.year >= 2022]

# THE FIX IS HERE: Filter by the list of London Borough codes
df_hpi_london = df_hpi_filtered[df_hpi_filtered['AreaCode'].isin(london_lad_codes)]

# --- 4. Engineer the Feature: Calculate Price Change by Borough ---
start_date = df_hpi_london['Date'].min()
end_date = df_hpi_london['Date'].max()

price_start = df_hpi_london[df_hpi_london['Date'] == start_date].set_index('AreaCode')['AveragePrice']
price_end = df_hpi_london[df_hpi_london['Date'] == end_date].set_index('AreaCode')['AveragePrice']

price_change = ((price_end - price_start) / price_start) * 100

# --- 5. Create the Final Aggregated HPI DataFrame ---
df_hpi_agg = pd.DataFrame({'HPI_Change_pct': price_change})
df_hpi_agg.dropna(inplace=True)
df_hpi_agg.reset_index(inplace=True)
df_hpi_agg.rename(columns={'AreaCode': 'ladcd'}, inplace=True) # Rename for the final merge

print("--- HPI Data Processed Successfully at the Borough Level ---")
print(df_hpi_agg.head())

# Define the output path for your HPI summary file
output_path = 'borough_level_hpi_summary.csv'

# Save the aggregated DataFrame to the new CSV
# index=False prevents pandas from writing the DataFrame index as a column
df_hpi_agg.to_csv(output_path, index=False)

print(f"Aggregated HPI data saved to: {output_path}")
```
