```python
import pandas as pd

# --- 1. Load your single, comprehensive LSOA lookup file ---
lsoa_lookup_path = '/Users/callumhall/Documents/Dissertation/Datasets/PCD to LSOA to MSOA to LAD.csv'

# THE FIX IS HERE: Add the encoding='latin1' parameter
df_lookup = pd.read_csv(
    lsoa_lookup_path, 
    usecols=['ladnm', 'pcds'], 
    encoding='latin1'  # <--- ADD THIS
)

# --- 2. Create a list of all London Borough names ---
london_boroughs = [
    'Camden', 'Greenwich', 'Hackney', 'Hammersmith and Fulham', 'Islington', 'Kensington and Chelsea',
    'Lambeth', 'Lewisham', 'Southwark', 'Tower Hamlets', 'Wandsworth', 'Westminster',
    'Barking and Dagenham', 'Barnet', 'Bexley', 'Brent', 'Bromley', 'Croydon', 'Ealing', 'Enfield',
    'Haringey', 'Harrow', 'Havering', 'Hillingdon', 'Hounslow', 'Kingston upon Thames', 'Merton',
    'Newham', 'Redbridge', 'Richmond upon Thames', 'Sutton', 'Waltham Forest',
    'City of London'
]

# --- 3. Filter the lookup to keep only London postcodes ---
# Using .str.contains() is a robust way to find the London boroughs
df_london_postcodes = df_lookup[df_lookup['ladnm'].isin(london_boroughs)]

# --- 4. Create a Python 'set' of London postcodes for super-fast filtering ---
london_postcodes_set = set(df_london_postcodes['pcds'])

print(f"Created a reference list of {len(london_postcodes_set)} postcodes in London.")
```

```python
# --- 1. Load your large prescription data file ---
prescription_data_path = '/Users/callumhall/Downloads/epd_202504.csv'
df_prescriptions = pd.read_csv(prescription_data_path)

# --- 2. Filter the DataFrame using the set of London postcodes ---
df_london_prescriptions = df_prescriptions[df_prescriptions['POSTCODE'].isin(london_postcodes_set)]

# --- 3. Save the filtered data to a new CSV ---
output_path = 'london_prescription_data.csv'
df_london_prescriptions.to_csv(output_path, index=False)

print(f"\nFiltered London prescription data has been saved to: {output_path}")
```

```python
import pandas as pd

# Load your saved London prescription data
df_london_prescriptions = pd.read_csv('london_prescription_data.csv')

# Ensure the BNF code column is a string to allow for filtering
df_london_prescriptions['BNF_CHAPTER_PLUS_CODE'] = df_london_prescriptions['BNF_CHAPTER_PLUS_CODE'].astype(str)

# --- THE FIX IS HERE ---
# Create a more robust filter that checks for multiple possible formats

# Format 1: With leading zero (e.g., '0401' for anxiolytics, '0403' for antidepressants)
is_anxiolytic_v1 = df_london_prescriptions['BNF_CHAPTER_PLUS_CODE'].str.startswith('0401')
is_antidepressant_v1 = df_london_prescriptions['BNF_CHAPTER_PLUS_CODE'].str.startswith('0403')

# Format 2: Without leading zero (e.g., '401', '403')
# We add a check to make sure we don't accidentally include Chapter 40, etc.
is_anxiolytic_v2 = df_london_prescriptions['BNF_CHAPTER_PLUS_CODE'].str.match(r'^401')
is_antidepressant_v2 = df_london_prescriptions['BNF_CHAPTER_PLUS_CODE'].str.match(r'^403')

# Combine all conditions. This finds rows that match ANY of these patterns.
df_mh_prescriptions = df_london_prescriptions[
    is_anxiolytic_v1 | is_antidepressant_v1 | is_anxiolytic_v2 | is_antidepressant_v2
]

# --- The rest of the code is the same ---

# Group by the drug description and sum the total number of items
top_mh_items = df_mh_prescriptions.groupby('BNF_DESCRIPTION')['ITEMS'].sum().reset_index()

# Sort the list to see the most common items at the top
top_mh_items_sorted = top_mh_items.sort_values(by='ITEMS', ascending=False)

# Display the top 20 most prescribed mental health items
print("--- Top 20 Most Prescribed Mental Health Items in London (Corrected) ---")
if top_mh_items_sorted.empty:
    print("Filter still resulted in an empty DataFrame. Please inspect the unique values of 'BNF_CHAPTER_PLUS_CODE'.")
else:
    print(top_mh_items_sorted.head(20))
```

```python
import pandas as pd

# Load your saved London prescription data
df_london_prescriptions = pd.read_csv('london_prescription_data.csv')

# Ensure the description column is a string and convert to lowercase for an easy search
df_london_prescriptions['BNF_DESCRIPTION'] = df_london_prescriptions['BNF_DESCRIPTION'].astype(str).str.lower()

# Find all rows where the drug description contains 'diazepam'
df_diazepam = df_london_prescriptions[df_london_prescriptions['BNF_DESCRIPTION'].str.contains('diazepam')]

# --- THE DIAGNOSIS ---
# Print the code and description for these rows
# This will reveal the true format of the anxiolytic codes in your file
print("--- Found Diazepam Prescriptions ---")
if df_diazepam.empty:
    print("No Diazepam found. Try another common anxiolytic name like 'lorazepam'.")
else:
    # We also need to see the BNF_CHAPTER_PLUS_CODE to confirm our original assumption was wrong
    print(df_diazepam[['BNF_CODE', 'BNF_CHAPTER_PLUS_CODE', 'BNF_DESCRIPTION']].head(10))
```

```python
# --- THE FIX: Use the 'BNF_CODE' column for specific filtering ---
# Ensure the 'BNF_CODE' column is a string to use .str methods
df_london_prescriptions['BNF_CODE'] = df_london_prescriptions['BNF_CODE'].astype(str)

# Filter for codes starting with '0401' (Anxiolytics) or '0403' (Antidepressants)
is_anxiolytic = df_london_prescriptions['BNF_CODE'].str.startswith('0401')
is_antidepressant = df_london_prescriptions['BNF_CODE'].str.startswith('0403')

df_mh_prescriptions = df_london_prescriptions[is_anxiolytic | is_antidepressant]

# --- The rest of the analysis now works correctly ---

# Group by the drug description and sum the total number of items
top_mh_items = df_mh_prescriptions.groupby('BNF_DESCRIPTION')['ITEMS'].sum().reset_index()

# Sort the list to see the most common items at the top
top_mh_items_sorted = top_mh_items.sort_values(by='ITEMS', ascending=False)

# Display the top 20 most prescribed mental health items
print("--- Top 20 Most Prescribed Mental Health Items in London (Corrected) ---")
if top_mh_items_sorted.empty:
    print("Filter still resulted in an empty DataFrame. The issue is more complex.")
else:
    print(top_mh_items_sorted.head(20))

# Create your aggregated summary for linking to LSOAs
df_practice_mh_summary = df_mh_prescriptions.groupby(['PRACTICE_CODE', 'POSTCODE'])['ITEMS'].sum().reset_index()
df_practice_mh_summary.rename(columns={'ITEMS': 'Total_MH_Items'}, inplace=True)
```

```python
# Define the output path for your new summary file
output_path = 'practice_level_mh_summary.csv'

# Save the aggregated DataFrame to the new CSV
# index=False prevents pandas from writing the DataFrame index as a column
df_practice_mh_summary.to_csv(output_path, index=False)

print(f"\nAggregated mental health prescription data saved to: {output_path}")
```
