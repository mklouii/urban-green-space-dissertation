```python
# --- 1. Define File Paths ---
samhi_path = '/Users/callumhall/Downloads/SAMHI Dataset.csv'  # <-- UPDATE THIS
lsoa_lookup_path = '/Users/callumhall/Documents/Dissertation/Datasets/master_lookup_2021_based.csv' # <-- UPDATE THIS

# --- 2. Load the SAMHI data, selecting only the most recent year ---
# We only need the 2011 LSOA code and the 2022 SAMHI index score
cols_to_use = ['lsoa11', 'samhi_index.2022']
df_samhi = pd.read_csv(samhi_path, usecols=cols_to_use)

# --- 3. Rename the columns for clarity ---
df_samhi.rename(columns={'lsoa11': 'LSOA_Code_2011', 'samhi_index.2022': 'SAMHI_Score'}, inplace=True)

# --- 4. Map to 2021 LSOA Codes ---
# This makes the SAMHI data compatible with your main project
df_lookup = pd.read_csv(lsoa_lookup_path, usecols=['lsoa11cd', 'LSOA21CD'])
df_samhi_mapped = pd.merge(
    df_samhi,
    df_lookup,
    left_on='LSOA_Code_2011',
    right_on='lsoa11cd',
    how='inner'
)

# --- 5. Create the final, clean DataFrame ---
# Keep only the 2021 LSOA code and the score
df_samhi_final = df_samhi_mapped[['LSOA21CD', 'SAMHI_Score']]

# --- 6. Drop Duplicates ---
df_samhi_final.drop_duplicates(subset=['LSOA21CD'])

# --- 6. Save the Final Cleaned File ---
output_path = 'new_cleaned_samhi_data.csv'
df_samhi_final.to_csv(output_path, index=False)

print(f"Cleaned and mapped SAMHI data saved to: {output_path}")
print("\n--- Verification ---")
print(df_samhi_final.head())
print(df_samhi.info())
```
