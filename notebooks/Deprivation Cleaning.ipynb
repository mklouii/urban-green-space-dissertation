```python
import pandas as pd

# File paths from your prompt
imd_path = '/Users/callumhall/Documents/Dissertation/Datasets/Cleaned Data/cleaned_IMD_data_2019.csv'
lookup_path = '/Users/callumhall/Documents/Dissertation/Datasets/master_lookup_2021_based.csv'

# Load the datasets
df_imd = pd.read_csv(imd_path)
df_lookup = pd.read_csv(lookup_path)
```

```python
# Check the column names in your lookup table
print("Columns in the lookup table:")
print(df_lookup.columns)
print("-" * 30)

# ==> IMPORTANT: Replace 'LSOA11CD' and 'LSOA21CD' below with the actual
# ==> column names for the 2011 and 2021 codes from your lookup file.

# Merge the IMD data with the lookup table
# This adds the 2021 LSOA codes to your IMD data based on the 2011 codes
updated_df = pd.merge(
    df_imd,
    df_lookup[['lsoa11cd', 'LSOA21CD']], # Select only the key columns from the lookup
    left_on='LSOA_Code_2011',           # Column in your IMD data
    right_on='lsoa11cd',                # Column in your lookup data
    how='left'                          # Keep all rows from the IMD data
)

# Display the first few rows to see the new 2021 column
print("DataFrame after merging:")
display(updated_df.head())
```

```python
# Drop the old 2011 code columns ('LSOA_Code_2011' from the original file
# and 'LSOA11CD' from the lookup file)
updated_df = updated_df.drop(columns=['LSOA_Code_2011', 'lsoa11cd'])

# Rename the 2021 code column to 'LSOA_Code'
# ==> IMPORTANT: Replace 'LSOA21CD' if your column name is different.
updated_df = updated_df.rename(columns={'LSOA21CD': 'LSOA_Code'})

# Display the final result to confirm the changes
print("Final DataFrame with updated and renamed LSOA column:")
display(updated_df.head())
```

```python
# Define the output path for the new file
output_path = '/Users/callumhall/Documents/Dissertation/Datasets/Cleaned Data/updated_IMD_data_2019.csv'

# Save to CSV, excluding the pandas index column
updated_df.to_csv(output_path, index=False)

print(f"Successfully saved updated data to {output_path}")
```
