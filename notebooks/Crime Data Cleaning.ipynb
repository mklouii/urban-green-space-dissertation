```python
import pandas as pd

# --- 1. Generate the list of all 12 monthly file paths ---
base_path = '/Users/callumhall/Documents/Dissertation/Archive/Gentrification/Datasets/Street Crime Data/'
# Create a date range for the 12 months you want to load
date_range = pd.date_range(start='2024-05-01', end='2025-04-30', freq='MS')

# Generate the file paths based on the year and month
crime_file_paths = []
for date in date_range:
    year = date.year
    month = f"{date.month:02d}" # Formats the month with a leading zero, e.g., '05'
    file_path = f"{base_path}{year}-{month}-metropolitan-street.csv"
    crime_file_paths.append(file_path)

print("Generated the following file paths to load:")
for path in crime_file_paths:
    print(path)

# --- 2. Load and combine the files into a single DataFrame ---
monthly_dfs = []
for filename in crime_file_paths:
    try:
        df = pd.read_csv(filename)
        monthly_dfs.append(df)
    except FileNotFoundError:
        print(f"\nWarning: Could not find file {filename}. It will be skipped.")

if not monthly_dfs:
    print("\nError: No crime data files were loaded. Please check your base_path and file names.")
else:
    df_crime_combined = pd.concat(monthly_dfs, axis=0, ignore_index=True)
    print(f"\nSuccessfully combined {len(monthly_dfs)} monthly files.")

    # --- 3. Select useful columns and aggregate by LSOA ---
    cols_to_keep = ['LSOA code', 'Crime type']
    df_crime_slim = df_crime_combined[cols_to_keep].copy()
    
    # Create a summary DataFrame with a total crime count
    df_crime_agg = df_crime_slim.groupby('LSOA code').size().reset_index(name='Total_Crimes_12m')
    
    print("\n--- Aggregated Crime Data Ready for Merging ---")
    print(df_crime_agg.head())
    
    # --- 4. Save the final aggregated file ---
    output_path = 'lsoa_level_crime_summary.csv'
    df_crime_agg.to_csv(output_path, index=False)
    print(f"\nFinal crime summary saved to: {output_path}")
```
